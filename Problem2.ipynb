```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
# capturing video with default webcam
cap = cv2.VideoCapture(0)
# time measure for fps
previous_time = time.perf_counter()


while cap.isOpened():
    # retrieve frame and response
    ret, frame = cap.read()
    if not ret:
        break
    # grayscale
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # canny edges
    edges = cv2.Canny(gray_frame, 25, 50)
    edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

    # edge highlight
    dimmed_frame = (frame * 0.3).astype(np.uint8)
    dimmed_frame[edges > 0] = [255, 255, 255]

    # edge enhancement
    edge_enhancement_factor = 0.4

    enhanced_frame = cv2.addWeighted(frame, 1-edge_enhancement_factor, edges_bgr, edge_enhancement_factor, 0)
    # concatenated frames with labels
    frames = cv2.hconcat([frame, dimmed_frame, enhanced_frame])
    cv2.putText(frames, "Original", (10,30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 1)
    cv2.putText(frames, "Masked", (650,30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 1)
    cv2.putText(frames, "Edge-Enhanced", (1290,30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 1)

    current_time = time.perf_counter()
    elapsed = current_time - previous_time
    fps = 1 / elapsed
    previous_time = current_time

    cv2.putText(frames, f"{fps:.1f} FPS", (10,450), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 1)

    cv2.imshow("Frames", frames)
    
    #cv2.imshow("Original", frame)
    #cv2.imshow("Masked", dimmed_frame)
    #cv2.imshow("Edge-Enhanced", enhanced_frame)


    # script stop when hit q
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
